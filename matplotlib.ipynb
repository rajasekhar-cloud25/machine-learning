## Matplotlib workflow

```python
# prepare data

x = [1, 2, 3, 4]
y = [1, 3, 6, 7]

# setup plot

fig, ax = plt.subplots(figsize=(5, 5)) # 5, 5 width and height

# plot data
ax.plot(x, y)

# customize plot
ax.set(title = "Simple Plot",
       xlabel = "x-axis",
       ylabel = "y-axis")

# save and show
fig.savefig("simple_plot.png")
fig.show()
```

```python
# make a plot from dictionary
# fig is the entire figure (the overall canvas).
# ax is the plotting area where you draw your graph.

# For a line plot: ax.plot(x, y)
# For a scatter plot: ax.scatter(x, y)
# For a vertical bar chart: ax.bar(x, height)
# For a horizontal bar chart: ax.barh(y, width)

nut_butter_prices = {"Almond butter": 10,
                     "Peanut butter": 8,
                     "Cashew butter": 12}



fig, ax = plt.subplots()

# create a bar chart
ax.bar(x = nut_butter_prices.keys(), height = nut_butter_prices.values());

ax.set(title = "Nut Butter Prices",
       xlabel = "Butter Types",
       ylabel = "Price ($)")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.barh(nut_butter_prices.keys(), nut_butter_prices.values());

plt.show()
```

```python
# Make some data for histograms and plot it

x = np.random.rand(1000)

fig, ax = plt.subplots()

ax.hist(x);
```

```python
# Two options for subplots

# subplot option 1

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows = 2,
                                             ncols = 2,
                                             figsize = (8, 5),
                                             )

# plot to each diffrent axis

ax1.plot(x, x / 2);
ax1.set(title = "Line Plot");

ax2.scatter(np.random.rand(10), np.random.rand(10));
ax2.set(title = "Scatter Plot");

ax3.bar(np.random.rand(10), np.random.rand(10));
ax3.set(title = "Bar Chart");

ax4.hist(np.random.rand(10));
ax4.set(title = "Histogram");

plt.tight_layout()
```

```python
# Subplots option 2

fig, ax = plt.subplots(nrows = 2,
                      ncols = 2,
                      figsize = (10, 5))

# Plot to each diffrent index
ax[0, 0].plot(x, x /2);
ax[0, 1].scatter(np.random.randint(10), np.random.randint(10));
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1, 1].hist(np.random.rand(100));

plt.tight_layout()
```

```python
# Data cleaning is the process of identifying and correcting or remove errors, missing values, duplicates in a dataset to improve data quality.
car_sales["Price"] = car_sales["Price"].str.replace(r'[$,]', "", regex = True)
car_sales
```

```python
car_sales["Sale Date"] = pd.date_range("1/1/2020", periods=len(car_sales))
car_sales
```

```python
# cumsum() calculates the cumulative sum of values, where each result is the sum of the current value and all preceding values.
car_sales["Total Sales"] = car_sales["Price"].astype(float).cumsum()

car_sales
```

```python
# Plot the total sales

car_sales.plot(x = "Sale Date", y = "Total Sales");
```

```python
# Reassign price column to float
car_sales["Price"] = car_sales["Price"].astype(float)

car_sales.plot(x = "Odometer (KM)", y = "Price", kind = "scatter");
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50.head()
```

```python
over_50.plot(kind = 'scatter',
             x = 'age',
             y = 'chol',
             c = 'target');
```

```python
# OO method

fix, ax = plt.subplots(figsize = (10, 6))

over_50.plot(kind = 'scatter',
             x = 'age',
             y = 'chol',
             ax = ax);
```

```python
from matplotlib.lines import lineStyles
fig, ax = plt.subplots(figsize = (10, 6))

# plot the data

scatter = ax.scatter(x = over_50['age'],
                     y = over_50['chol'],
                     c = over_50['target'])


ax.set(title = "Heart Disease and Cholesterol Levels",
       xlabel = 'Age',
       ylabel = 'Cholesterol');


# add a legend
ax.legend(*scatter.legend_elements(), title = 'Target');

# add a horizontal line

ax.axhline(over_50['chol'].mean(),
           linestyle = '--');
```

```python
over_50.head()
```

```python
fig, (ax0, ax1) = plt.subplots(nrows = 2,
                               ncols = 1,
                               figsize = (6, 10),
                               sharex = True)

scatter = ax0.scatter(x = over_50['age'],
                      y = over_50['chol'],
                      c = over_50['target']);

ax0.set(title = "Heart Disease and Cholesterol Levels",
        ylabel = 'Cholesterol');

ax0.legend(*scatter.legend_elements(), title = 'Target');

ax0.axhline(over_50['chol'].mean(),
            linestyle = '--');

scatter = ax1.scatter(x  = over_50['age'],
                      y = over_50['thalach'],
                      c = over_50['target']);


ax1.set(title = "Heart Disease and Max Heart Rate",
        xlabel = 'Age',
        ylabel = 'Max Heart Rate');

ax1.legend(*scatter.legend_elements(), title = 'Target');

ax1.axhline(y = over_50['thalach'].mean(),
            linestyle = '--');

fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = 'bold');

plt.tight_layout();
```
