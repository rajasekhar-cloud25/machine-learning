## Matplotlib workflow

```python
# prepare data

x = [1, 2, 3, 4]
y = [1, 3, 6, 7]

# setup plot

fig, ax = plt.subplots(figsize=(5, 5)) # 5, 5 width and height

# plot data
ax.plot(x, y)

# customize plot
ax.set(title = "Simple Plot",
       xlabel = "x-axis",
       ylabel = "y-axis")

# save and show
fig.savefig("simple_plot.png")
fig.show()
```

```python
# make a plot from dictionary
# fig is the entire figure (the overall canvas).
# ax is the plotting area where you draw your graph.

# For a line plot: ax.plot(x, y)
# For a scatter plot: ax.scatter(x, y)
# For a vertical bar chart: ax.bar(x, height)
# For a horizontal bar chart: ax.barh(y, width)

nut_butter_prices = {"Almond butter": 10,
                     "Peanut butter": 8,
                     "Cashew butter": 12}



fig, ax = plt.subplots()

# create a bar chart
ax.bar(x = nut_butter_prices.keys(), height = nut_butter_prices.values());

ax.set(title = "Nut Butter Prices",
       xlabel = "Butter Types",
       ylabel = "Price ($)")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.barh(nut_butter_prices.keys(), nut_butter_prices.values());

plt.show()
```

```python
# Make some data for histograms and plot it

x = np.random.rand(1000)

fig, ax = plt.subplots()

ax.hist(x);
```

```python
# Two options for subplots

# subplot option 1

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(nrows = 2,
                                             ncols = 2,
                                             figsize = (8, 5),
                                             )

# plot to each diffrent axis

ax1.plot(x, x / 2);
ax1.set(title = "Line Plot");

ax2.scatter(np.random.rand(10), np.random.rand(10));
ax2.set(title = "Scatter Plot");

ax3.bar(np.random.rand(10), np.random.rand(10));
ax3.set(title = "Bar Chart");

ax4.hist(np.random.rand(10));
ax4.set(title = "Histogram");

plt.tight_layout()
```

```python
# Subplots option 2

fig, ax = plt.subplots(nrows = 2,
                      ncols = 2,
                      figsize = (10, 5))

# Plot to each diffrent index
ax[0, 0].plot(x, x /2);
ax[0, 1].scatter(np.random.randint(10), np.random.randint(10));
ax[1, 0].bar(nut_butter_prices.keys(), nut_butter_prices.values());
ax[1, 1].hist(np.random.rand(100));

plt.tight_layout()
```
