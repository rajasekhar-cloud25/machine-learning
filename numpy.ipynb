## Aggregation

## Taking many values in an array and summarizing them into fewer values

Common NumPy aggregation functions
```
np.sum(a)	Sum of values
np.mean(a)	Average
np.min(a)	Smallest value
np.max(a)	Largest value
np.std(a)	Standard deviation
np.var(a)	Variance
np.median(a)	Middle value
np.argmin(a)	Index of smallest value
np.argmax(a)	Index of largest value
```
